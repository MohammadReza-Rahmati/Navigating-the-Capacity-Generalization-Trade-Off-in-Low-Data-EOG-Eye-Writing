```python
import numpy as np
from scipy import stats

def confidence_interval_95(scores):
  
    n = len(scores)
    mean = np.mean(scores)
    se = stats.sem(scores)
    t_crit = stats.t.ppf(0.975, df=n-1)
    ci = t_crit * se
    return mean, ci

def wilcoxon_signed_rank_test(scores_a, scores_b, alpha=0.05):
  
    stat, p_value = stats.wilcoxon(scores_a, scores_b, alternative='two-sided')
    significant = p_value < alpha
    diff = np.mean(scores_a) - np.mean(scores_b)
    return stat, p_value, significant, diff

# LOSO (6 fold)
loso_conv1d       = [0.766667, 0.983333, 0.639344, 0.586777, 0.816667, 0.785124]
loso_conv1d_fast  = [0.841667, 0.966667, 0.704918 , 0.694215, 0.691667, 0.851240]
loso_conv1d_faster= [0.658333, 0.916667, 0.696721, 0.586777, 0.708333 , 0.702479 ]

# CI
print("=" * 65)
print("Confidence Intervals (95%) - LOSO Protocol (6 folds)")
print("=" * 65)
for name, scores in [("conv1d", loso_conv1d),
                      ("conv1d_fast", loso_conv1d_fast),
                      ("conv1d_faster", loso_conv1d_faster)]:
    mean, ci = confidence_interval_95(scores)
    print(f"{name:15s}: {mean:.4f} ± {ci:.4f}")

# Wilcoxon
print("\n" + "=" * 65)
print("Wilcoxon Signed-Rank Tests - LOSO Protocol")
print("=" * 65)
comparisons = [
    ("conv1d vs conv1d_fast",    loso_conv1d, loso_conv1d_fast),
    ("conv1d vs conv1d_faster",  loso_conv1d, loso_conv1d_faster),
    ("conv1d_fast vs faster",    loso_conv1d_fast, loso_conv1d_faster),
]

for name, a, b in comparisons:
    w, p, sig, diff = wilcoxon_signed_rank_test(a, b)
    print(f"{name:30s}: Δ={diff:+.4f}, W={w:.2f}, p={p:.4f}, "
          f"{'✓ Significant' if sig else '✗ Not significant'}")
```
