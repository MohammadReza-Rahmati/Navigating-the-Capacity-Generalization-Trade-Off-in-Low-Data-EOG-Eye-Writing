```python
import numpy as np
from scipy import stats

def confidence_interval_95(scores):
   
    n = len(scores)
    mean = np.mean(scores)
    se = stats.sem(scores)
    t_crit = stats.t.ppf(0.975, df=n-1)
    ci = t_crit * se
    return mean, ci

def wilcoxon_signed_rank_test(scores_a, scores_b, alpha=0.05):
   
    stat, p_value = stats.wilcoxon(scores_a, scores_b, alternative='two-sided')
    significant = p_value < alpha
    diff = np.mean(scores_a) - np.mean(scores_b)
    return stat, p_value, significant, diff


conv1d_scores        = [0.950000, 0.958333, 0.786885, 0.842975, 0.916667, 0.966942]
conv1d_fast_scores   = [0.808333, 0.833333, 0.573770, 0.553719, 0.708333, 0.851240]
conv1d_faster_scores = [0.766667, 0.775000, 0.500000, 0.421488, 0.533333, 0.851240]

# Confidence Intervals
print("=" * 60)
print("Confidence Intervals (95%)")
print("=" * 60)
for name, scores in [("conv1d", conv1d_scores),
                      ("conv1d_fast", conv1d_fast_scores),
                      ("conv1d_faster", conv1d_faster_scores)]:
    mean, ci = confidence_interval_95(scores)
    print(f"{name:15s}: {mean:.4f} ± {ci:.4f}")

# Wilcoxon signed-rank tests
print("\n" + "=" * 60)
print("Wilcoxon Signed-Rank Tests (User-Dependent)")
print("=" * 60)
comparisons = [
    ("conv1d vs conv1d_fast",    conv1d_scores, conv1d_fast_scores),
    ("conv1d vs conv1d_faster",  conv1d_scores, conv1d_faster_scores),
    ("conv1d_fast vs faster",    conv1d_fast_scores, conv1d_faster_scores),
]

for name, a, b in comparisons:
    w, p, sig, diff = wilcoxon_signed_rank_test(a, b)
    print(f"{name:30s}: Δ={diff:+.4f}, W={w:.2f}, p={p:.4f}, "
          f"{'✓ Significant' if sig else '✗ Not significant'}")
```
