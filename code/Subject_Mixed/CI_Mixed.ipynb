```python
import numpy as np
from scipy import stats

def confidence_interval_95(scores):

    n = len(scores)
    mean = np.mean(scores)
    se = stats.sem(scores)
    t_crit = stats.t.ppf(0.975, df=n-1)
    ci = t_crit * se
    return mean, ci



# Subject-Mixed (5 fold) 

mixed_conv1d        = [0.94, 0.93, 0.93, 0.91, 0.90] # mean=0.9220, std=0.0065
mixed_conv1d_fast   = [0.86, 0.84, 0.87, 0.85, 0.84]  # mean=0.8520, std=0.0018
mixed_conv1d_faster = [0.82, 0.80, 0.83, 0.81, 0.83]  # mean=0.8180, std=0.0014


print("=" * 70)
print("Confidence Intervals (95%) - Subject-Mixed Protocol (5 folds)")
print("=" * 70)
for name, scores in [("conv1d", mixed_conv1d),
                      ("conv1d_fast", mixed_conv1d_fast),
                      ("conv1d_faster", mixed_conv1d_faster)]:
    mean, ci = confidence_interval_95(scores)
    print(f"{name:15s}: {mean:.4f} ± {ci:.4f}")

```
